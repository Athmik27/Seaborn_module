```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")

sns.violinplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Type2",
    linewidth=3,
    linecolor="cyan",
    order=["Bug","Water","Fire","Grass"]
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")

sns.violinplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Type2",
    linewidth=3,
    linecolor="cyan",
    order=["Bug","Water","Fire","Grass"],
    split="True" #  splits each violin into two halves, one for each hue category.
)

plt.show()
```
