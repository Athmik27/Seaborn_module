```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

var_name = pd.read_csv("sample.csv") 

sns.heatmap(var_name.select_dtypes(include="number"),vmin=0,vmax=12) 
# They control the range of values used for the heatmap's color scale.
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

var_name = pd.read_csv("sample.csv") 

sns.heatmap(var_name.select_dtypes(include="number"),vmin=0,vmax=12,cmap="ocean_r") 
# we use cmap for changing the color of the heatmap
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

var_name = pd.read_csv("sample.csv") 

sns.heatmap(var_name.select_dtypes(include="number"),vmin=0,vmax=12,cmap="ocean_r",annot=True) 
# this annot=True gives you the numbers inside the heatmap
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create our own numerical data
var_name = np.linspace(1, 10, 10).reshape(2, 5)

print(var_name)
sns.heatmap(
    var_name,
    vmin=0,
    vmax=12,
    cmap="ocean_r",
    annot=True
)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Create our own numerical data
var_name = np.linspace(1, 10, 10).reshape(2, 5)

print(var_name)
variable=sns.heatmap(
    var_name,
    vmin=0,
    vmax=12,
    cmap="ocean_r",
    annot=True
)
variable.set(xlabel="Python",ylabel="number")
plt.show()
```
