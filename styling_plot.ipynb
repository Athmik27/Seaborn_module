```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
sns.barplot(x="Type1",y="Name",data=var_name) 
# if we need to remove axis spines
# despine() removes some of the axis spines.
# axis spines is are those which form border of graph
sns.despine()
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# if we need to increase the size for figure / scale and content
plt.Figure(figsize=(25,3))
sns.barplot(x="Type1",y="Name",data=var_name) 
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")

sns.set_style("white")
# if we need to chamge the context of the graph we use sns.set_context
sns.set_context("poster",font_scale=0.5) #"poster" this gives the size of poster instead we can also use "paper" etc
sns.barplot(x="Type1",y="Name",data=var_name) 

plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")

sns.set_style("white")
sns.set_context("poster",font_scale=0.5) 
# we need to change the color of the graph we use palette
sns.barplot(x="Type1",y="Name",data=var_name,palette="winter") 

plt.grid()
plt.show()
```
