```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv")



sns.stripplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue='Type2',
   edgecolor='Red',
    linewidth=1,
    jitter=0.2 ,
    marker="<" # this sets the maker shape to the specified one
    )
plt.show()
```

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv")



sns.stripplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue='Type2',
   edgecolor='Red',
    linewidth=1,
    jitter=0.2 ,
    marker="<",
    alpha=0.5 # this tells the transperancy of the marker color.
    )
plt.show()
```
