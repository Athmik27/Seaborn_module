```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv")

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv").head(5) # .head(5) means only first 5 rows

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv").head(5)

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Height" # we'll now differentiate based on the requirements so we use hue=" column name"
)

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv").head(5)

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Height",
    palette="viridis" # now if we need to change the color we use palette=" color name "
)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv").head(5)

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Height",
    palette="viridis" ,
    height=5 # we use this as if we need to get the required height of the graph.
)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

var_name = pd.read_csv("sample.csv").head(5)

print(var_name)

g = sns.catplot(
    x="Type1",
    y="Weight",
    data=var_name,
    hue="Height",
    palette="viridis" ,
    height=5 ,
    kind="bar" # we use kind as if we need bar graph to be represented inside the cat-plot.
)# in kind="  " we can give strip,swarm,box,violin,boxen,point,bar,count.
plt.show()
```
