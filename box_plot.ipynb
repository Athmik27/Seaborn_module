```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Read the CSV file
data = pd.read_csv("sample.csv")

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create box plot
sns.boxplot(
    x="Type1",
    y="Weight",
    data=data
)

# Add labels and title
plt.xlabel("Pokemon Type")
plt.ylabel("Weight")
plt.title("Weight Distribution by Pokemon Type")

# Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Read the CSV file
data = pd.read_csv("sample.csv")

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create box plot
sns.boxplot(
    x="Type1",
    y="Weight",
    data=data,
    hue="Type1",
    
   order=["Bug","Water","Fire","Grass"],# order is used to set the formay in which we want to arrange
   showmeans="True" # usually the normal graph shows us the median but showmeans shows the average/mean.



)
   

# Add labels and title
plt.xlabel("Pokemon Type")
plt.ylabel("Weight")
plt.title("Weight Distribution by Pokemon Type")

# Display the graph
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Read the CSV file
data = pd.read_csv("sample.csv")

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create box plot
sns.boxplot(
    x="Type1",
    y="Weight",
    data=data,
    hue="Type1",
    
   order=["Bug","Water","Fire","Grass"],# order is used to set the formay in which we want to arrange
   showmeans="True",
    linewidth=3,
     palette='plasma' # usually the normal graph shows us the median but showmeans shows the average/mean.



)
   

# Add labels and title
plt.xlabel("Pokemon Type")
plt.ylabel("Weight")
plt.title("Weight Distribution by Pokemon Type")

# Display the graph
plt.show()
```
