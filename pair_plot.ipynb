```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
  # No  Name  Type1   Type2  Height  Weight  Legendary all these are the column so if we need any selected column we use
# vars=["",""]
# pairplot() works mainly with numerical columns
sns.pairplot(var_name,hue="Type1",vars=["Height","Weight"])

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# if we need to control the order of the categories/colors created by hue we use hue_order.

sns.pairplot(var_name,hue="Type1",vars=["Height","Weight"],hue_order=["Bug","Fire","Water","Grass"])

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# if we need to change the color we use palette=" "
sns.pairplot(var_name,
             hue="Type1",
             vars=["Height","Weight"],
             hue_order=["Bug","Fire","Water","Grass"],
             palette="ocean_r")

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# if we need to change the marker we use 
sns.pairplot(var_name,
             hue="Type1",
             vars=["Height","Weight"],
             hue_order=["Bug","Fire","Water","Grass"],
             palette="ocean_r",
             markers=["*",">"])

plt.show()
```
