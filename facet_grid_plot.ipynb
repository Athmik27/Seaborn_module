```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# here fg is an name of an variable
fg=sns.FacetGrid(var_name)
# fg.map(plot_function, x_column, y_column) 
# map() tells us to use the plot_function and plot the graph.
fg.map(plt.scatter,"Height","Weight") 
# plt.scatter we use because we need to plot an garph using scatter plot.(any plot we can use)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
# by help of facet we can draw an multiple plots of graph we use col="".
fg=sns.FacetGrid(var_name,col="Type1")
fg.map(plt.scatter,"Weight","Height") 
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
fg=sns.FacetGrid(var_name,col="Type1",hue="Name")# hue → keeps them in the same plot but uses different colors
fg.map(plt.scatter,"Weight","Height") 
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var_name = pd.read_csv("sample.csv")
fg = sns.FacetGrid(var_name, col="Type2", hue="Type1")
fg.map(plt.scatter, "Weight", "Height")
fg.add_legend()

plt.show()
```
